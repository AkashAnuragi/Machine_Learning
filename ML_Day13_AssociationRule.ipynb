#### Association Rules
Association rules is an <b>Unsupervised ML Technique</b> used to discover interesting relationships or patterns between items in a dataset<br>
<b>Example :-</b> It answers the question: <b>"If a customer buys item A, what i sthe chance that they will also buy item B"</b>

<b>Real Life Dataset Example</b>

    <tr><th>Customer</th> <th>Purchased</th></tr>

After Analysis , we got the rule<br><b>Rule :</b>  Bread -> Butter<br>
Customer who buy <b>Bread</b> often buy <b>Butter</b> as well.

#### Important Terms
1. <b>Support</b>
    Support tells us <b>how frequently am itemset apperas in the entire dataset</b><br>
    Formula:<br>
    Support(A->B) = Transaction containing A and B / Total Transaction<br>
    Support = 3/5 = 60%

2. <b>Confidence</b>
    Confidence tells us <b>If A occurs, how likely B to occur?</b><br>
    Formula:<br>
    Confidence = Support(A->B)/Support(A)<br>
    Confidence = 3/4 = 75%<br>
    75% of customers who buy Bread also buy Butter

3. <b>Lift</b>
    Lift measures <b>how strong the relationship is</b> between two items. <br>
    Formula:<br>
    Lift = Confidence/Support(B)<br>
    Lift = 75 / 60 =><b>1.25</b><br>
    - Lift = 1 (No Relationship)
    - Lift < 1 (Negative Relation)
    - Lift > 1 (Positive Relation)
