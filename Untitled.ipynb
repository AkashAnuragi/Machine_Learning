### Reinforcement Learning
 In this learning we create an agent who take action in an environment and  we'll give it reward  and penalty

 - agent - ROBOT
 - Environment = Area where agent will do action
 - State - Current Position
 - Action - Robot moves/will perform a task
 - Reward/Penalty - Result (and we will store it in Q-table)
 - Goal - Reach on that state (where we wanna go)
 

```python
import random
import numpy as np
```

```python
goal = 5
state = 0
```

```python
# basic Simple Python Program (No learning)
print("Starting Stage :", state)
while state != goal:
    action = 1 # Move forward
    # Next state
    state = state + action
    # Reward
    if state == goal:
        reward = 100
    else:
        reward = -1
    print("Current State ",state, " and reward ",reward)
print("Reached Goal!")    
```

```python
# better but no Learning
print("Starting Stage :",state)
while state != goal:
    action = np.random.choice([1,-1]) # Random Action
    # Next state
    new_state = state + action

    if new_state <0:
        new_state = 0
    if new_state >goal:
        new_state = goal
    state = new_state
    # Reward
    if state == goal:
        reward = 100
    else:
        reward = -1
    print("Current State ",state, " and reward ", reward)
print("Reached Goal")
```

```python
Q_table = np.array([0,0,0,0,0,100])
```

```python
for i in range(goal):
    state = 0
    goal = 5
    print("Starting Stage :",state)
    while state!=goal:
        action = 1  # Move forward
        #Next state
        next_state = state+action
        Q_table[state] = Q_table[next_state]-1
        state = next_state
        # Reward 
        if state == goal:
            reward = 100
        else:
            reward = -1
        print("Current State ",state, " and reward ",reward)
    print("Reached Goal!")
```
